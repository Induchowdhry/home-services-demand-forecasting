# Milestone 1 — Data Exploration

This notebook inspects the supplied training data and verifies whether it matches the home-services demand-forecasting project brief.

```python
import pandas as pd
from pathlib import Path

DATA = Path('..') / 'data' / 'aiml_training_data.csv'
df = pd.read_csv(DATA)
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
df.info()
```

```python
print('Missing values:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

```python
print('Intent counts:')
print(df['intent'].value_counts())
print('\nConfidence summary:')
print(df['confidence'].describe())
```

## Dataset validation conclusion

The supplied CSV has `text`, `intent`, and `confidence` only. It does not contain the booking time series required by the demand-forecasting brief (date, city, service type, weather, and demand/bookings). Therefore no forecasting MAPE is calculated from this file.

```python
required = {'date', 'city', 'service_type', 'bookings'}
missing = required - set(df.columns)
print('Forecasting columns missing from supplied CSV:', sorted(missing))
```
